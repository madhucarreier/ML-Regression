```python
#importing the libirary for this Module Program
import matplotlib as plt
import pandas as pd
```

```python
#read the data from CSV file and store the value dataset variable
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset
```

```python
#ordinal data convert into true or fakse
dataset=pd.get_dummies(dataset)
```

```python
dataset
```

```python
#input  field assign into indep and output field assign into from dep
indep=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
dep=dataset['charges']
```

```python
#train test split here we will pass the indep and dep  parameter then split X_train,y_train input and test data 
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(indep, dep, test_size = 1/3,random_state = 0)
```

```python
#preprocessing tool in machine learning that standardizes numeric features
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
#supervised machine learning algorithm that builds an ensemble of multiple decision trees

from sklearn.ensemble import RandomForestRegressor
```

```python

from sklearn.model_selection import GridSearchCV
param_grid = {'criterion':['squared_error', 'absolute_error'],
'max_features': ['sqrt', 'log2', None],
'n_estimators':[10,100]}

#defined parameter values for the RF Regressor model




```

```python
#GridSearchCV pass the RandomForestRegressor duntion,refit find the best result verbose is full debugging mode in most software and programming tools
grid = GridSearchCV(RandomForestRegressor(), param_grid, refit=True, verbose=3, n_jobs=-1)
```

```python
grid.fit(X_train, y_train)  #Model is trained and build
```
