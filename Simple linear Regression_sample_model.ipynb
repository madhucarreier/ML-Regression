```python
## declare the library and read the adta from the csv file
import pandas as pd
dataset=pd.read_csv("Salary_Data.csv")
dataset
```

```python
## after read the file data store indepent varaiable
indepent
```

```python
##Read the spectific column value
indepent=dataset[["yearsofexperience"]]
```

```python
##Read the spectific output column value
dependent=dataset[["salary"]]
```

```python
dependent
```

```python
##split the dta and declare the inpu and output varable  x is input variable and y is output variable
## sklearn .model used to split the input and output varaible
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(indepent, dependent, test_size=0.30)
```

```python
X_train  #only input features input
```

```python
 y_train  # only output values to train
```

```python
## sklearn.linear_model used to create the LinearRegression model   
# Learning phase
from sklearn.linear_model import LinearRegression
##regressor=LinearRegression()- non parameteris function
regressor=LinearRegression()
## fit used used to create AI brain model
regressor.fit(X_train,y_train)
```

```python
regressor
```

```python
#weight value  coef_ nothing thera is declare the value
weight=regressor.coef_
weight
```

```python
##bais term of your trained model
bais=regressor.intercept_
bais
```

```python
y_pred=regressor.predict(X_test) X_test actual output is y_test
```

```python

##measure how well your machine learning model performs if the score nearly by 1 good value or
##if the value nearl y by zero data is not perforly well

from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred)
print(score)
```
